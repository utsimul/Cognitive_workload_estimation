# Exploratory Data Analysis for EDA features

EDA = electrodermal activity (skin conductance). The main idea is that sweat glands are controlled only by the sympathetic nervous system (high stress = high sympathetic nervous system activation). 

EDA Signal
   =
Tonic Component   +   Phasic Component
(slow baseline)       (fast reactions)

*Tonic component* = background arousal level (overall nervous system activation)
This changes slowly over minutes.

If there is more variation in tonic component (high std dev) then it means that the brain states are undergoing transitions. 

*Phasic component*
short bursts of sympathetic activation. (sudden difficulty, decision making, surprise, errors)

*SCR peaks* Number of detected SCR peaks in window Difficult tasks have many SCR peaks.

Note that SCR peaks and EDA in general is not just for stress - it is for sympathetic arousal which could be due to attention, novelty, etc.

```python
import os
import pickle
import pandas as pd


base_path = "/Users/krisha/Projs/Neuro-brain-states/Datasets/EPIStress"

patient_ids = ["140", "141", "247", "257", "260","313","362","404","437","446","491"]


tasks = {
    "arithmetix_easy": "easy",
    "arithmetix_hard": "hard",
    "relaxation_video": "relaxation"
}

feature_columns = [
    "SCR_Peaks_N",
    "SCR_Peaks_Amplitude_Mean",
    "EDA_Tonic_Mean",
    "EDA_Tonic_Std",
    "EDA_Phasic_Mean",
    "EDA_Phasic_Std"
]

def read_pickle(path):
    with open(path, "rb") as f:
        return pickle.load(f)

all_data = []

for patient_id in patient_ids:

    for task_folder, task_label in tasks.items():

        path = os.path.join(
            base_path,
            "ES"+patient_id,
            "Lab",
            "Features",
            task_folder,
            "EDA_features_30_sec.pickle"
        )

        
        if not os.path.exists(path):
            print(f"Missing: {path}")
            continue

        
        df = read_pickle(path).copy()

       
        df = df[feature_columns].copy()

       
        df["task_label"] = task_label
        df["patient_id"] = patient_id

        all_data.append(df)

        print(
            f"{patient_id} | {task_label} | "
            f"{df.shape[0]} windows"
        )


EDA_all = pd.concat(
    all_data,
    ignore_index=True
)

print("\nFinal shape:", EDA_all.shape)

EDA_all.head()
```
