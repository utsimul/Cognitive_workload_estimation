# Exploratory Data Analysis for EDA features

EDA = electrodermal activity (skin conductance). The main idea is that sweat glands are controlled only by the sympathetic nervous system (high stress = high sympathetic nervous system activation). 

EDA Signal
   =
Tonic Component   +   Phasic Component
(slow baseline)       (fast reactions)

*Tonic component* = background arousal level (overall nervous system activation)
This changes slowly over minutes.

If there is more variation in tonic component (high std dev) then it means that the brain states are undergoing transitions. 

*Phasic component*
short bursts of sympathetic activation. (sudden difficulty, decision making, surprise, errors)

*SCR peaks* Number of detected SCR peaks in window Difficult tasks have many SCR peaks.

Note that SCR peaks and EDA in general is not just for stress - it is for sympathetic arousal which could be due to attention, novelty, etc.

# EDA - combining all subjects

```python
import os
import pickle
import pandas as pd


base_path = "/Users/krisha/Projs/Neuro-brain-states/Datasets/EPIStress"

patient_ids = ["140", "141", "247", "257", "260","313","362","404","437","446","491"]


tasks = {
    "arithmetix_easy": "easy",
    "arithmetix_hard": "hard",
    "relaxation_video": "relaxation"
}

feature_columns = [
    "SCR_Peaks_N",
    "SCR_Peaks_Amplitude_Mean",
    "EDA_Tonic_Mean",
    "EDA_Tonic_Std",
    "EDA_Phasic_Mean",
    "EDA_Phasic_Std"
]

def read_pickle(path):
    with open(path, "rb") as f:
        return pickle.load(f)

all_data = []

for patient_id in patient_ids:

    for task_folder, task_label in tasks.items():

        path = os.path.join(
            base_path,
            "ES"+patient_id,
            "Lab",
            "Features",
            task_folder,
            "EDA_features_30_sec.pickle"
        )

        
        if not os.path.exists(path):
            print(f"Missing: {path}")
            continue

        
        df = read_pickle(path).copy()

       
        df = df[feature_columns].copy()

       
        df["task_label"] = task_label
        df["patient_id"] = patient_id

        all_data.append(df)

        print(
            f"{patient_id} | {task_label} | "
            f"{df.shape[0]} windows"
        )


EDA_all = pd.concat(
    all_data,
    ignore_index=True
)

print("\nFinal shape:", EDA_all.shape)

EDA_all.head()
```

```python
outlier_summary = []

for feature in EDA_features:

    Q1 = EDA_all[feature].quantile(0.25)
    Q3 = EDA_all[feature].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (EDA_all[feature] < lower_bound) |
        (EDA_all[feature] > upper_bound)
    )

    outlier_summary.append({
        "feature": feature,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "n_outliers": outliers.sum(),
        "outlier_percentage": 100 * outliers.mean()
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

```python
patient_outliers = []

for feature in EDA_features:

    Q1 = EDA_all[feature].quantile(0.25)
    Q3 = EDA_all[feature].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    temp = EDA_all.copy()

    temp["is_outlier"] = (
        (temp[feature] < lower) |
        (temp[feature] > upper)
    )

    summary = (
        temp
        .groupby("patient_id")["is_outlier"]
        .mean()
        .reset_index()
    )

    summary["feature"] = feature

    patient_outliers.append(summary)

patient_outliers = pd.concat(
    patient_outliers,
    ignore_index=True
)

patient_outliers.head()
```

## Kruskal Wallis

```python
from scipy.stats import kruskal

kruskal_results = []

for feature in EDA_features:

    groups = [
        group[feature].dropna().values
        for _, group in patient_task_means.groupby("task_label")
    ]

    statistic, p_value = kruskal(*groups)

    kruskal_results.append({
        "feature": feature,
        "H_statistic": statistic,
        "p_value": p_value
    })

kruskal_results = pd.DataFrame(
    kruskal_results
)

kruskal_results
```

```python
from scipy.stats import mannwhitneyu

task_pairs = [
    ("easy", "hard"),
    ("easy", "relaxation"),
    ("hard", "relaxation")
]

pairwise_results = []

for feature in EDA_features:

    for task_1, task_2 in task_pairs:

        x = patient_task_means.loc[
            patient_task_means["task_label"] == task_1,
            feature
        ].dropna()

        y = patient_task_means.loc[
            patient_task_means["task_label"] == task_2,
            feature
        ].dropna()

        statistic, p_value = mannwhitneyu(
            x,
            y,
            alternative="two-sided"
        )

        pairwise_results.append({
            "feature": feature,
            "comparison": f"{task_1} vs {task_2}",
            "U_statistic": statistic,
            "p_value": p_value,
            "n_1": len(x),
            "n_2": len(y)
        })

pairwise_results = pd.DataFrame(
    pairwise_results
)

pairwise_results
```

```python
def rank_biserial_correlation(x, y):

    U, _ = mannwhitneyu(
        x,
        y,
        alternative="two-sided"
    )

    n1 = len(x)
    n2 = len(y)

    rbc = (2 * U) / (n1 * n2) - 1

    return rbc
```

```python
pairwise_results = []

for feature in EDA_features:

    for task_1, task_2 in task_pairs:

        x = patient_task_means.loc[
            patient_task_means["task_label"] == task_1,
            feature
        ].dropna()

        y = patient_task_means.loc[
            patient_task_means["task_label"] == task_2,
            feature
        ].dropna()

        statistic, p_value = mannwhitneyu(
            x,
            y,
            alternative="two-sided"
        )

        effect_size = rank_biserial_correlation(x, y)

        pairwise_results.append({
            "feature": feature,
            "comparison": f"{task_1} vs {task_2}",
            "U_statistic": statistic,
            "p_value": p_value,
            "effect_size": effect_size,
            "n_1": len(x),
            "n_2": len(y)
        })

pairwise_results = pd.DataFrame(
    pairwise_results
)

pairwise_results
```

## Benjamin Hochberg

```python
from statsmodels.stats.multitest import multipletests

reject, p_fdr, _, _ = multipletests(
    pairwise_results["p_value"],
    method="fdr_bh"
)

pairwise_results["p_fdr"] = p_fdr
pairwise_results["significant_fdr"] = reject
```

## Aggregate => 1 row for the entire task duration (like the input to the model)

```python
import pandas as pd


feature_columns = [
    "SCR_Peaks_N",
    "SCR_Peaks_Amplitude_Mean",
    "EDA_Tonic_Mean",
    "EDA_Tonic_Std",
    "EDA_Phasic_Mean",
    "EDA_Phasic_Std"
]


EDA_task_level = (
    EDA_all
    .groupby(["patient_id", "task_label"])[feature_columns]
    .agg([
        "mean",
        "std",
        "median",
        "min",
        "max",
        lambda x: x.quantile(0.25),
        lambda x: x.quantile(0.75)
    ])
)


EDA_task_level.columns = [
    f"{feature}_{stat}"
    for feature, stat in zip(
        EDA_task_level.columns.get_level_values(0),
        [
            "mean",
            "std",
            "median",
            "min",
            "max",
            "q25",
            "q75"
        ] * len(feature_columns)
    )
]

EDA_task_level = EDA_task_level.reset_index()

print(EDA_task_level.shape)
EDA_task_level.head()
```

```python
normality_results = []

for feature in feature_cols:

    stat,p = shapiro(EDA_task_level[feature])

    normality_results.append([
        feature,
        stat,
        p
    ])

normality = pd.DataFrame(
    normality_results,
    columns=[
        "Feature",
        "Shapiro Statistic",
        "p-value"
    ]
)

normality
```

```python
levene_results = []

for feature in feature_cols:

    easy = EDA_task_level.loc[
        EDA_task_level.task_label=="easy",
        feature
    ]

    hard = EDA_task_level.loc[
        EDA_task_level.task_label=="hard",
        feature
    ]

    relax = EDA_task_level.loc[
        EDA_task_level.task_label=="relaxation",
        feature
    ]

    stat,p = levene(easy,hard,relax)

    levene_results.append([
        feature,
        stat,
        p
    ])

levene_df = pd.DataFrame(
    levene_results,
    columns=[
        "Feature",
        "Statistic",
        "p-value"
    ]
)

levene_df
```

```python
results = []

for feature in feature_cols:

    easy = EDA_task_level.loc[
        EDA_task_level.task_label=="easy",
        feature
    ]

    hard = EDA_task_level.loc[
        EDA_task_level.task_label=="hard",
        feature
    ]

    relax = EDA_task_level.loc[
        EDA_task_level.task_label=="relaxation",
        feature
    ]

    stat,p = kruskal(
        easy,
        hard,
        relax
    )

    results.append([
        feature,
        stat,
        p
    ])

results = pd.DataFrame(
    results,
    columns=[
        "Feature",
        "H",
        "p-value"
    ]
)

results["adj_p"] = multipletests(
    results["p-value"],
    method="fdr_bh"
)[1]

results.sort_values("adj_p")
```

```python
comparisons = [
    ("easy","hard"),
    ("easy","relaxation"),
    ("hard","relaxation")
]

pairwise = []

for feature in feature_cols:

    for g1,g2 in comparisons:

        x = EDA_task_level.loc[
            EDA_task_level.task_label==g1,
            feature
        ]

        y = EDA_task_level.loc[
            EDA_task_level.task_label==g2,
            feature
        ]

        stat,p = mannwhitneyu(
            x,
            y,
            alternative="two-sided"
        )

        pairwise.append([
            feature,
            g1,
            g2,
            stat,
            p
        ])

pairwise = pd.DataFrame(
    pairwise,
    columns=[
        "Feature",
        "Group1",
        "Group2",
        "Statistic",
        "p-value"
    ]
)

pairwise["adj_p"] = multipletests(
    pairwise["p-value"],
    method="fdr_bh"
)[1]

pairwise.sort_values("adj_p")
```

```python
results["eta_squared"] = (
    results["H"] - 3 + 1
) / (
    len(EDA_task_level) - 3
)

results.sort_values(
    "eta_squared",
    ascending=False
)
```
